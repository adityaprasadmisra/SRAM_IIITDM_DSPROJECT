```python
# ==========================================
# STEP 4: PREPARE BASELINE DATASET
# ==========================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("../data/raw/sram_fault_dataset.csv")
print("Dataset shape:", df.shape)
# Target
y = df["fault_class"].copy()

# Columns excluded due to leakage, identifiers,
# simulation outcomes, or redundancy
excluded_columns = [
    "sample_id",
    "corner_id",
    "fault_class",
    "fault_name",
    "fault_target",
    "severity_value",
    "converged",
    "snm_asym_hold_v",
    "snm_drop_v",
    "iddq_norm",
    "t_write_norm",
    "dv_bl_norm",
    "snm_hold_norm",
    "snm_read_norm",
    "read_write_delay_ratio",
    "snm_ratio",
    "dv_bl_vdd_ratio"
]

# Candidate input features
X = df.drop(columns=excluded_columns, errors="ignore").copy()

# Remove constant columns automatically
constant_columns = [
    col for col in X.columns
    if X[col].nunique(dropna=False) <= 1
]

X.drop(columns=constant_columns, inplace=True)

print("Feature matrix:", X.shape)
print("Target matrix:", y.shape)

print("\nSelected features:")
print(X.columns.tolist())

print("\nConstant columns removed:")
print(constant_columns)

print("\nMissing values:")
print(X.isnull().sum().sum())
```

```python
print("Numerical features:")
print(X.select_dtypes(include=np.number).columns.tolist())

print("\nCategorical features:")
print(X.select_dtypes(include=["object", "category"]).columns.tolist())
```

```python
# Combine selected features and target
baseline_df = X.copy()
baseline_df["fault_class"] = y

# Save
baseline_df.to_csv(
    "../data/sram_baseline_dataset.csv",
    index=False
)

print("Baseline dataset saved successfully!")
```
